## Introduction to Pandas

**Pandas** is a powerful open-source Python library used for data manipulation and analysis. It provides flexible data structures, such as DataFrames and Series, making it easy to clean, explore, and analyze structured data efficiently.

Below are some basics of using pandas for data analysis, including creating DataFrames, selecting data, and performing basic operations.

```python
import pandas as pd
```

```python
fruits = pd.Series(['apple', 'banana', 'cherry', 'date', 'Mango', 'orange', 'pear'])
fruits
```

```python
boooks = pd.Series(['The Great Gatsby', 'To Kill a Mockingbird', '1984', 'Pride and Prejudice', 'The Catcher in the Rye'])
boooks
```
